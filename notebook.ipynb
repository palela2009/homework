```
birth_year = 2009
name = "ალექსანდრე"
surname = "ფალელაშვილი"
current_year = "2026"

age = int(current_year) - birth_year
sentence = f"მე {name} {surname} დავიბადე {birth_year} წელს, შესაბამისად ვარ {age} წლის."

print(sentence)
```

```
yes_votes = 119
no_votes = 82

total_votes = yes_votes + no_votes
yes_percentage = (yes_votes / total_votes) * 100
no_percentage = (no_votes / total_votes) * 100

print(f"YES: {yes_votes} = {yes_percentage:.2f}%")
print(f"NO: {no_votes} = {no_percentage:.2f}%")
```
